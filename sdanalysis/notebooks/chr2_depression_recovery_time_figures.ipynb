```python
import sys
import os
import pandas as pd
notebook_dir = os.getcwd()
sys.path.insert(0, os.path.abspath(os.path.join(notebook_dir, '..')))  # Add the project root directory to the path
import custom_io as cio
import env_reader
import data_documentation
import matplotlib.pyplot as plt
from scipy.stats import circstd
import seaborn as sns
```

```python
save_figs = True
merge_jrgeco = True  # merge red-shifted chr2 + jrgeco recordings with chr2 + gcamp recordings to have a single category "chr2_szsd" (or chr2_sd)
```

```python
fpath_baseline_darkest = cio.open_file("Choose baseline minus darkest results excel file (e.g. bl-to-darkest-point_....xlsx)")
fpath_recovery_time = cio.open_file("Choose recovery time results excel file (e.g. recovery_times_....xlsx)")
```

```python
df_baseline_darkest = pd.read_excel(fpath_baseline_darkest)
df_recovery_time = pd.read_excel(fpath_recovery_time)
```

```python
df_baseline_darkest["bl-darkest_normalized"] = df_baseline_darkest["bl-darkest"]/df_baseline_darkest["baseline"]
```

```python
# limit to chr2_(sz)sd and jrgeco_(sz)sd categories. Also limit to CA1, although no NC chr2 or jrgeco yet
df_to_plot_bl_darkest = df_baseline_darkest[(df_baseline_darkest['exp_type'].isin(["chr2_sd", "jrgeco_sd", "chr2_szsd", "jrgeco_szsd"])) & (df_baseline_darkest["win_type"] == "CA1")]
df_to_plot_recovery_time = df_recovery_time[(df_recovery_time['exp_type'].isin(["chr2_sd", "jrgeco_sd", "chr2_szsd", "jrgeco_szsd"])) & (df_recovery_time["window_type"] == "CA1")]

if merge_jrgeco:
    df_to_plot_bl_darkest.loc[df_to_plot_bl_darkest['exp_type'] == "jrgeco_sd", 'exp_type'] = "chr2_sd"
    df_to_plot_bl_darkest.loc[df_to_plot_bl_darkest['exp_type'] == "jrgeco_szsd", 'exp_type'] = "chr2_szsd"
    df_to_plot_recovery_time.loc[df_to_plot_recovery_time['exp_type'] == "jrgeco_sd", 'exp_type'] = "chr2_sd"
    df_to_plot_recovery_time.loc[df_to_plot_recovery_time['exp_type'] == "jrgeco_szsd", 'exp_type'] = "chr2_szsd"

fig, axs = plt.subplots(1, 2, figsize=(16, 8))
sns.violinplot(data=df_to_plot_bl_darkest, x='exp_type', y='bl-darkest_normalized', hue="exp_type", fill=False, inner="point", ax=axs[0])
axs[0].set_title("Baseline - Darkest Point post-stim")
axs[0].set_ylabel("Baseline - Darkest Point post-stim")
axs[0].set_xlabel("Experiment Type")

sns.violinplot(data=df_to_plot_recovery_time, x='exp_type', y='t_recovery_s', hue="exp_type", fill=False, inner="point", ax=axs[1])
axs[1].set_title("Recovery Time")
axs[1].set_ylabel("Recovery Time (s)")
axs[1].set_xlabel("Experiment Type")

if save_figs:
    plt.savefig(os.path.join("D:\\Output\\", "optostim_baseline_darkest-normalized_recovery_time.pdf"))
plt.show()
```
